```python
import pandas as pd
df = pd.read_excel('file1/students.xlsx', index_col='번호')
df['합계'] = df['국어'] + df['영어'] + df['수학']
df['평균'] = df['합계'] / 3
df['등수'] = df['합계'].rank(ascending=False, method='min').astype(int)
# df['등수'] = df['합계'].rank(ascending=False, method='dense').astype(int)
df.sort_values(by='등수', ascending=True, inplace=True)
df.head(30)
```
