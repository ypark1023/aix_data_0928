```python
import pandas as pd
df = pd.read_excel('file1/scores1.xlsx', index_col='지원번호')
df
```

```python
import random

df['요일'] = 0

for i in range(8):
    df.loc[df.index[i], '요일'] = random.randint(0, 6)


def d_change(dx):
    if dx == 1:
        return '월'
    elif dx == 2:
        return '화'
    elif dx == 3:
        return '수'
    elif dx == 4:
        return '목'
    elif dx == 5:
        return '금'
    elif dx == 6:
        return '토'
    else:
        return '일'

df['요일'] = df['요일'].apply(d_change)
df
```
