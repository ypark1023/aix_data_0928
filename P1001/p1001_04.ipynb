```python
import pandas as pd
df = pd.read_excel('file1/scores1.xlsx', index_col='지원번호')
df['합계'] = df['국어'] + df['영어'] + df['수학'] + df['과학'] + df['사회']
df
```

```python
# def lowChange(s):
#     if pd.notnull(s):
#         return str(s).lower()
#     return '없음'

# df['SW특기'] = df['SW특기'].apply(lowChange)
# df

def add5(row):
    # row['SW특기']가 null이 아니면 합계에 5를 더함, 아니면 기존 합계 유지
    if pd.notnull(row['SW특기']):
        return row['합계'] + 5
    return row['합계']


# axis=1을 유지하여 각 행별로 함수 적용
df['합계'] = df.apply(add5, axis=1)
df
```
