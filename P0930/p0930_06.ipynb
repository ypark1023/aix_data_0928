### 문자열 함수
#### str.upper(), str.lower(): 대/소문자 변환
#### str.len(): 문자열 길이 계산
#### str.contains(): 특정 문자열 포함 여부 확인
#### str.replace(): 문자열 치환
#### str.split(): 문자열 분할
#### str.strip(), str.lstrip(), str.rstrip(): 공백 제거
#### str.extract(): 정규 표현식으로 문자열 추출
#### str.cat(): 문자열 결합
#### unique(), nunique(): 고유 값 처리
#### isnull(), fillna(): 널 값 처리
