```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus'] = False
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
```

```python
df = pd.read_excel('file1/students.xlsx', index_col='번호')
# df
```

```python
# sizes = np.random.rand(100)*1000
sizes = df['학년']*500
```

### Scatter 그래프

```python
plt.figure(figsize=(10,10))
# plt.scatter(df['영어'], df['수학'], s=sizes, c=df['학년'], alpha=0.5)
plt.scatter(df['영어'], df['수학'], s=sizes, c=df['학년'], alpha=0.5, cmap='cool')
plt.xlabel('영어')
plt.ylabel('수학')
plt.colorbar(ticks=[1,2,3], label='학년', shrink=0.5, orientation='horizontal')
plt.show()
```
