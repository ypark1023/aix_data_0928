```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus'] = False
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
```

```python
df = pd.read_csv('file1/drinks.csv')
df['total_servings'] = df['beer_servings']+df['spirit_servings']+df['wine_servings']
# # df.groupby('continent')['total_servings'].sum().sort_values(ascending=False)
# drk = df['total_servings'].sort_values(ascending=False).head(20)
# # drk = df.groupby('continent')['total_servings'].sum().sort_values(ascending=False)
# drk
# x = drk.index
# y = drk.values

sort_df = df.sort_values('total_servings', ascending=False)
x = sort_df['country']
y = sort_df['total_servings']
# plt.bar(x,y)

# colors = ['red' if country == 'South Korea' else 'lightgray' for country in x]


k_sort_df = sort_df[sort_df['country']=='South Korea']
x2 = k_sort_df['country']  # 한국이름
y2 = k_sort_df['total_servings'] #한국합계
plt.figure(figsize=(20, 5))
plt.bar(x, y, color= 'lightgrey', label='국가별')
plt.bar(x2,y2,color='r', label='한국') #한국그래프
plt.xlabel('Country', color='red', loc='right')
plt.ylabel('Total', color='blue', loc='top')
plt.xticks(rotation=90)
plt.legend()

plt.show()
```
