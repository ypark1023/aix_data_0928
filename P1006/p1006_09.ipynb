```python
df = pd.read_csv('file0/namu_data.cvs')
df2 = df.head(5)
df2
```

```python
def text_cleaning(text):
    if pd.isna(text):
        return ''
    
    pattern = re.compile(r'[^ 가-힣a-zA-Z0-9]+')
    return pattern.sub('', text)

# df['content']
# text_cleaning(df['content'][0])

df2['title'] = df['title'].apply(lambda x : text_cleaning(x))
df2['category'] = df['category'].apply(lambda x : text_cleaning(x))
df2['content'] = df['content'].apply(lambda x : text_cleaning(x))
df2
```

```python
titles = " ".join(df2['title'].tolist())
categorys = " ".join(df2['category'])
contents = " ".join(df2['content'])

contents
```

```python
stopwords = ['상위', '문서', '내용', '누설', '아래', '해당', '설명', '표기', '추가', '모든', '사용', '매우', '가장',
            '줄거리', '요소', '상황', '편집', '틀', '경우', '때문', '모습', '정도', '이후', '사실', '생각', '인물',
            '이름', '년월']

count3 = Counter({x:count2[x] for x in count2 if x not in stopwords})
count3
```

```python
df3 = (
    pd.DataFrame(count3.items(), columns=['word', 'count'])
    .sort_values('count', ascending=False)
    .reset_index(drop=True)
)
df3
```
