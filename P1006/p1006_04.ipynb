```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus'] = False
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
```

```python
df = pd.read_csv('file1/chipotle.tsv', sep='\t')
df
```

```python
chip = df.groupby('item_name')['quantity'].sum().sort_values(ascending=False)
x = chip.index
y = chip.values

plt.figure(figsize=(10,5))
plt.xticks(rotation=90)
plt.bar(x,y)
plt.show()
```

```python
chip = df.groupby('item_name')['quantity'].sum().sort_values(ascending=False)
chip2 = chip[df.groupby('item_name')['quantity'].sum()>=100]
x = chip.index
y = chip.values
x2 = chip2.index
y2 = chip2.values

plt.figure(figsize=(10,5))
plt.xticks(rotation=90)
plt.bar(x,y)
plt.bar(x2,y2, label = '100개 이상')
plt.legend()
plt.show()
```
