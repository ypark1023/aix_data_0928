```python
# SELENIUM으로 열기
df = pd.read_csv('file0/namu1.csv')
data = []
for url in df['url']:
    namu_url = url
    # options = Options()
    # options.add_experimental_option("excludeSwitches", ["enable-automation"])
    # options.add_experimental_option("useAutomationExtension", False)
    # options.add_argument("--disable-blink-features=AutomationControlled")
    # browser = webdriver.Chrome(options=options)
    # browser.maximize_window()
    # browser.get(namu_url)
    # time.sleep(2)

    res = requests.get(url)
    time.sleep(1)
    soup = BeautifulSoup(res.text, 'lxml')
    title0 = soup.find('a',{'class':'J23-5aq4'})
    title = title0.get_text(strip=True) if title0 else None
    category0 = soup.find('ul',{'class':'EVpyuXop'})
    category = category0.get_text(strip=True) if category0 else None
    content0 = soup.find('div',{'class':'_6YZgjHNT Oud4u5fe'})
    content = content0.get_text(strip=True)  if content0 else None
    data.append({
        'title' : title,
        'category' : category,
        'content' : content 
    })

n_df = pd.DataFrame(data)
n_df
# n_df.to_csv('file0/namu_data.cvs')
```
