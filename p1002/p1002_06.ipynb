```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus'] = False
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
```

```python
df = pd.read_excel('file1/scores1.xlsx', index_col='지원번호')
df
```

```python
# x = [1,2,3]
# x = df.index
x = df['이름'].to_list()
y1 = np.array(df['국어'])
y2 = np.array(df['영어'])
y3 = np.array(df['수학'])
clrs = ['r', 'g', 'b']

plt.figure(figsize=(8,5))
plt.title('학생 성적 그래프')


plt.bar(x,y1, label='국어', width=0.5)
plt.bar(x,y2,bottom=y1, label='영어', width=0.5)
plt.bar(x,y3,bottom=y1+y2, label='수학', width=0.5)
plt.grid(axis='y', linestyle=':', alpha=0.7)

# for i, v in enumerate(y1):
#     plt.text(x[i], y1[i]+1, v, ha='center')
# for i, v in enumerate(y2):
#     plt.text(x[i], y1[i]+y2[i]+1, v, ha='center')
# for i, v in enumerate(y3):
#     plt.text(x[i], y1[i]+y2[i]+y3[i]+1, v, ha='center')


for i, (v, v2, v3) in enumerate(zip(y1, y2, y3)):
    plt.text(x[i], v/2, v, ha='center', va='center')
    plt.text(x[i], v+v2/2, v, ha='center', va='center')
    plt.text(x[i], v+v2+v3/2, v, ha='center', va='center')


plt.yticks([0,50,100,150,200,250,300,350])
plt.legend(ncol = 3)
plt.tight_layout()
plt.show()
```
