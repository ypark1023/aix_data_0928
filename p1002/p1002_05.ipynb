```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus'] = False
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
```

```python
x = [1,2,3]
y = [10,5,30]

plt.figure(figsize=(10,5), dpi=200, facecolor='gray') #기본 dpi는 70?
plt.plot(x, y, label='Dataset1', linewidth=3, marker='o', color = 'blue')
plt.legend()
plt.title('Graph')
plt.xlabel("Month", color='red', loc='right')
plt.ylabel("Nums", color='blue', loc='top')

plt.xticks([1,1.5,2,2.5,3])
# plt.yticks([-30,-20,-10,0,10,20,30])

# plt.text(x[0], y[0]+1, 10, ha='center', color = 'red')
# plt.text(x[1], y[1]+1, 5, ha='center')
# plt.text(x[2], y[2]+1, 30, ha='center')

for i, v in enumerate(y):
    plt.text(x[i], y[i]+1, y[i], ha='center', color='red')


# plt.plot(x, y)
# plt.savefig('file1/aaa.png', dpi=30)
# plt.savefig('file1/aab.png', dpi=200)
plt.show()
```
