```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus'] = False
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
```

```python
import pandas as pd
import numpy as np
df = pd.read_csv('file1/chipotle.tsv', sep='\t')
df
```

```python
# df.groupby('item_name').count()
df['item_name'].value_counts().head(10)
# df.groupby('item_name')['quantity'].sum().sort_values(ascending=False)
df.groupby('item_name')['quantity'].sum().sort_values(ascending=False).head(10)
# df.groupby('item_name')['quantity'].sum().sort_values(ascending=False).head(10).to_list()
# plt.plot(x,y)
# plt.show()
```

```python
# top3 = (
#     df.groupby('item_name')['quantity']
#     .sum()
#     .sort_values(ascending=False)
#     .head(3)
# )

# x = top3.index  # 아이템 이름들
# y = top3.values  # 수량 합계들

top10 = (
    df.groupby('item_name')['quantity']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

x = top10.index  # 아이템 이름들
y = top10.values  # 수량 합계들

plt.bar(x, y)
# x축 이름이 길어서 겹치는 것을 방지하기 위해 회전 및 정렬 설정
plt.xticks(rotation=45, ha='right')

plt.title('Top Items by Quantity')
plt.xlabel('Item Name')
plt.ylabel('Quantity')
plt.tight_layout()  # 여백 조정
plt.show()
```
